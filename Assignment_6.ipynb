```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler

print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset downloaded successfully!")
```

```python
print("First Five Rows:")
display(df.head())

print("\nDataset Shape:", df.shape)
```

```python
print("Missing Values Before Processing:")
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Fare"] = df["Fare"].fillna(df["Fare"].median())

print("Missing numerical values handled successfully!")
```

```python
for col in ["Embarked", "Sex"]:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing categorical values handled successfully!")
```

```python
df = df.drop(columns=["Cabin"])

print("Cabin column removed successfully!")
```

```python
print("Missing Values After Processing:")
print(df.isnull().sum())
```

```python
encoder = LabelEncoder()

df["Sex_Encoded"] = encoder.fit_transform(df["Sex"])

print("Sex column encoded successfully!")

display(df[["Sex", "Sex_Encoded"]].head())
```

```python
df = pd.get_dummies(
    df,
    columns=["Embarked"],
    drop_first=True,
    dtype=int
)

print("Embarked column encoded successfully!")

display(df.head())
```

```python
scaler = MinMaxScaler()

columns_to_scale = ["Age", "Fare"]

df[columns_to_scale] = scaler.fit_transform(
    df[columns_to_scale]
)

print("Normalization completed successfully!")

display(df[["Age", "Fare"]].head())
```

```python
print("Final Processed Dataset:")
display(df.head())

print("\nFinal Dataset Shape:", df.shape)
```

```python
df.to_csv("preprocessed_titanic.csv", index=False)

print("Preprocessed dataset saved successfully!")
```
