```python
#handling outliers using IQR method(for year)
Q1 = df['Year'].quantile(0.25)
```

```python
Q3 = df['Year'].quantile(0.75)
```

```python
IQR = Q3-Q1
```

```python
#define oulier boundaries
lower_bound = Q1-1.5*IQR
upper_bound = Q3 + 1.5*IQR
```

```python
#cap year outliers
df['Year'] = np.where(df['Year']>upper_bound,upper_bound,
                      np.where(df['Year']<lower_bound,lower_bound,df['Year']))
```

4. Feature Scaling

```python
#using standard scaler on selling price and year
scaler = StandardScaler()
```

```python
#reshape required for fitting scaler
df[['Selling_Price','Year']] = scaler.fit_transform(df[['Selling_Price','Year']])
```
